```python
# Import learning curve utility from sklearn
from sklearn.model_selection import learning_curve

# Create figure with 2 subplots side by side, 16x6 inches
fig, ax = plt.subplots(1, 2, figsize=(16, 6))
# Adjust subplot spacing and margins
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)

# Loop through polynomial degrees 2 and 9
for i, degree in enumerate([2, 9]):
    # Generate learning curves using cross validation
    # Returns: training sizes (N), training scores (train_lc), validation scores (val_lc)
    N, train_lc, val_lc = learning_curve(
        # Create pipeline: polynomial features -> linear regression
        make_pipeline(PolynomialFeatures(degree), LinearRegression()),
        X, y,  # Input data and target values 
        cv=7,  # Use 7-fold cross validation
        train_sizes=np.linspace(0.3, 1, 25)  # Test 25 training sizes from 30% to 100%
    )

    # Plot training scores
    ax[i].plot(N, np.mean(train_lc, axis=1), 
               color='blue', 
               label='training score')
    
    # Plot validation scores  
    ax[i].plot(N, np.mean(val_lc, axis=1), 
               color='red', 
               label='validation score')
    
    # Add horizontal line at mean of final scores
    ax[i].hlines(np.mean([train_lc[-1], val_lc[-1]]), 
                 N[0], N[-1],  # Line spans full x-range
                 color='gray', 
                 linestyle='dashed')
    
    # Set plot styling
    ax[i].set_ylim(0, 1)  # Score range 0-1
    ax[i].set_xlim(N[0], N[-1])  # X-axis spans training sizes
    ax[i].set_xlabel('training size')
    ax[i].set_ylabel('score') 
    ax[i].set_title(f"degree = {degree}", size=14)
    ax[i].legend(loc='best')
    ax[i].margins(0.2)  # Add 20% padding around plot
    
```

We're using a technique called "Learning Curves" to understand how well our model learns and if we need more data or a different model complexity.


What are Polynomial Degrees?
-Think of polynomial degrees like this:
Degree 1: straight line (y = mx + b)
Degree 2: curved line (y = ax² + bx + c)
Degree 9: very wiggly line (y = ax⁹ + bx⁸ + ... + ix + j)

What is Cross Validation?
-It's like testing a student:
Split data into groups (folds)
Train on most groups
Test on the remaining group
Repeat with different test groups
Average the results
In our code, cv=7 means we do this 7 times with different splits.


Where can this be used?
-Customer Behavior
Input: past purchases, browsing history
Output: likelihood to buy
Determine if you have enough customer data

### L1 & L2 EXAMPLES


```python
# Import necessary libraries
from sklearn.linear_model import Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt

# Create sample data
np.random.seed(42)
X = np.random.randn(100, 20)  # 100 samples, 20 features
true_weights = np.array([1, 0.5, 0.2, 0.1] + [0] * 16)  # Only first 4 features are relevant
y = np.dot(X, true_weights) + np.random.randn(100) * 0.1

# Split and scale data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train models with different regularization strengths
alphas = [0.0001, 0.001, 0.01, 0.1, 1, 10]
l1_models = []
l2_models = []

for alpha in alphas:
    # L1 (Lasso)
    lasso = Lasso(alpha=alpha)
    lasso.fit(X_train_scaled, y_train)
    l1_models.append(lasso)
    
    # L2 (Ridge)
    ridge = Ridge(alpha=alpha)
    ridge.fit(X_train_scaled, y_train)
    l2_models.append(ridge)

# Plot coefficients
plt.figure(figsize=(15, 5))

# L1 (Lasso) plot
plt.subplot(121)
for model, alpha in zip(l1_models, alphas):
    plt.plot(model.coef_, label=f'alpha={alpha}')
plt.title('L1 (Lasso) Regularization')
plt.xlabel('Feature')
plt.ylabel('Coefficient Value')
plt.legend()

# L2 (Ridge) plot
plt.subplot(122)
for model, alpha in zip(l2_models, alphas):
    plt.plot(model.coef_, label=f'alpha={alpha}')
plt.title('L2 (Ridge) Regularization')
plt.xlabel('Feature')
plt.ylabel('Coefficient Value')
plt.legend()

plt.tight_layout()
plt.show()
```

### L1 (Лассо) Регуляризация:

Делает коэффициенты точно равными нулю (выбор признаков)
Хорошо, когда вы подозреваете, что многие признаки не имеют значения
Варианты использования:
Выбор генов в анализе ДНК
Выбор признаков в многомерных данных
Когда вам нужны разреженные модели

### L2 (Ридж) Регуляризация:

Сжимает коэффициенты до нуля, но не точно до нуля
Хорошо, когда многие признаки вносят небольшой вклад
Варианты использования:
Мультиколлинеарные данные (коррелированные признаки)
Когда вы хотите сохранить все признаки
Универсальная регуляризация

Аппроксимация - это приближение сложной функции более простой. Давайте разберем на простых примерах:


```python
# Import correct module
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Create the pipeline
def PolynomialRegression():
    return make_pipeline(PolynomialFeatures(), LinearRegression())

# Define parameter grid
param_grid = {
    'polynomialfeatures__degree': np.arange(21),
    'linearregression__fit_intercept': [True, False]
}

# Create and run GridSearchCV
grid = GridSearchCV(PolynomialRegression(),
                   param_grid, 
                   cv=7)
                   
```

```python
# Fit the grid search
grid.fit(X, y)
```

```python
grid.best_params_
{'linearregression__fit_intercept': False, 'polynomialfeatures__degree': 4}
```

```python
model = grid.best_estimator_

plt.scatter(X.ravel(), y)
lim = plt.axis()
y_test = model.fit(X, y).predict(X_test)
plt.plot(X_test.ravel(), y_test);
plt.axis(lim);
```
